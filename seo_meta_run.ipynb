# SEO Metadata Update Runner
Run the updated improve_meta.py script to apply the new local search metadata improvements, then inspect a sample HTML file for title and description changes.

```python
import os
from pathlib import Path

project_dir = Path(r"C:\Users\LOBO\Downloads\kaira-1.0.0")
os.chdir(project_dir)

print('Running improve_meta.py...')
exit_code = os.system('python improve_meta.py')
print('\nScript exit code:', exit_code)

sample_files = ['index.html', 'tshirts.html', 'all-products.html', 'why-us.html']
for fname in sample_files:
    path = project_dir / fname
    if path.exists():
        text = path.read_text(encoding='utf-8', errors='ignore')
        print(f'\n--- {fname} ---')
        for line in text.splitlines()[:40]:
            if '<title>' in line or '<meta name="description"' in line or '<meta name="keywords"' in line:
                print(line)
    else:
        print(f'{fname} not found')
```
